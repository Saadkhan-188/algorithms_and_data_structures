  # Question One: Check Duplicates in Tree

  Given the `root` of a binary tree, check whether it is contains a duplicate value. If a duplicate exists, return the duplicate value. If there are multiple duplicates, return the one with the closest distance to the root. If no duplicate exists, return -1.

  ## Examples

  Output: 2

  ### Example 2

  ![](./images/q1_ex2.png)

  Input: `root = [1, 10, 2, 3, 10, 12, 12]`

  Output: 10

  ### Example 3

  ![](./images/q1_ex3.png)

  Input: `root = [10, 9, 8, 7]`

  Output: -1

#### Starter Code for Question 1

```python
# Definition for a binary tree node.
from collections import defaultdict
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right
def is_duplicate(root: TreeNode) -> int:
    if not root:
        return 0

    # Dictionary to store hash -> count mappings
    hash_count = defaultdict(int)
    duplicate_count = 0

    def get_subtree_hash(node):
        """
        Generate a hash for the subtree rooted at node.
        Returns the hash and updates duplicate count.
        """
        nonlocal duplicate_count

        if not node:
            return hashlib.md5(b"null").hexdigest()

        # Get hashes for left and right subtrees
        left_hash = get_subtree_hash(node.left)
        right_hash = get_subtree_hash(node.right)

        # Create unique representation and hash it
        subtree_repr = f"{node.val}|{left_hash}|{right_hash}"
        subtree_hash = hashlib.md5(subtree_repr.encode()).hexdigest()

        # Count occurrences of this hash
        hash_count[subtree_hash] += 1

        # If we've seen this hash exactly twice, it's a new duplicate
        if hash_count[subtree_hash] == 2:
            duplicate_count += 1

        return subtree_hash

    get_subtree_hash(root)
    return duplicate_count

# Example usage and test cases
if __name__ == "__main__":
    # Test case 1: Tree with duplicate subtrees
    #       1
    #      / \
    #     2   3
    #    /   / \
    #   4   2   4
    #      /
    #     4
    root1 = TreeNode(1)
    root1.left = TreeNode(2)
    root1.left.left = TreeNode(4)
    root1.right = TreeNode(3)
    root1.right.left = TreeNode(2)
    root1.right.left.left = TreeNode(4)
    root1.right.right = TreeNode(4)
    
    print("Test case 1:")
    print(f"Number of duplicate subtrees: {is_duplicate(root1)}")  # Expected: 2
    
    # Test case 2: Tree with no duplicates
    root2 = TreeNode(1)
    root2.left = TreeNode(2)
    root2.left.left = TreeNode(4)
    root2.right = TreeNode(3)
    root2.right.right = TreeNode(5)
    
    print("Test case 2:")
    print(f"Number of duplicate subtrees: {is_duplicate(root2)}")  # Expected: 0
    
    # Test case 3: Tree with identical subtrees
    root3 = TreeNode(1)
    root3.left = TreeNode(2)
    root3.left.left = TreeNode(3)
    root3.right = TreeNode(2)
    root3.right.left = TreeNode(3)
    
    print("Test case 3:")
    print(f"Number of duplicate subtrees: {is_duplicate(root3)}")  # Expected: 2
```

# Note to self
## The solution uses a depth-first search (DFS) approach with backtracking:
Key concepts:

DFS traversal: We explore each path from root to leaf completely before moving to the next path
Backtracking: After exploring a subtree, we remove the current node from the path to avoid affecting other paths
Leaf detection: A node is a leaf if it has no left or right children

Algorithm steps:

Handle edge case where root is None
Use a helper function dfs that maintains the current path
For each node, add it to the current path
If it's a leaf, add a copy of the current path to results
Otherwise, recursively explore left and right children
Remove the current node from the path (backtrack)

Time complexity: O(N) where N is the number of nodes, since we visit each node once
Space complexity: O(H) for the recursion stack, where H is the height of the tree
The function returns all possible paths from the root to any leaf node as a list of lists.
Key concepts:

Serialization: Convert each subtree into a unique string representation
Pattern matching: Use a hash map to count occurrences of each subtree pattern
Duplicate detection: A subtree is considered duplicate when we've seen the same pattern at least twice

Algorithm steps:

Perform post-order traversal of the tree
For each node, serialize its subtree as "val,left_subtree,right_subtree"
Count occurrences of each serialized pattern
When a pattern appears for the second time, increment duplicate counter
Return the total count of duplicate subtree patterns

Key details:

Uses post-order traversal to build serialization from bottom up
Null nodes are represented as "null" in serialization
Only counts each duplicate pattern once (when count reaches exactly 2)
The serialization format ensures uniqueness for different tree structures

Time complexity: O(N) where N is the number of nodes
Space complexity: O(N) for the hash map storing serializations
The function returns the count of distinct duplicate subtree patterns found in the tree

Key improvements with hashing:

Better collision resistance: MD5/SHA-256 hashes virtually eliminate the risk of different subtrees having the same representation
Memory efficiency: Hash strings are fixed length regardless of subtree size
Performance: Hash computation is very fast, and hash comparisons are O(1)

Three versions provided:

is_duplicate(): Uses MD5 hashing - good balance of speed and collision resistance
is_duplicate_sha256(): Uses SHA-256 - maximum collision resistance for critical applications
is_duplicate_compact(): Uses truncated MD5 - most memory efficient

Algorithm benefits:

Scalability: Works efficiently even with very large trees or large node values
Robustness: Handles edge cases like negative numbers, large values, and complex tree structures
Consistent performance: Hash-based comparison is always O(1) regardless of subtree complexity

Time complexity: O(N) where N is the number of nodes
Space complexity: O(N) for storing hashes, but with much better constants than string-based approaches
The hashing approach is particularly beneficial when dealing with large trees or when node values are large numbers, as it avoids the potential issues with string concatenation and comparison that could arise in the non-hashed version.
The recursion works by:

Base case: If the node is None, return a hash for "null"
Recursive case: For each node, recursively get hashes for left and right children first, then combine them with the current node's value to create the subtree hash

This is a post-order traversal pattern - we process children before processing the current node, which is necessary because we need the hashes of the subtrees before we can create the hash for the current subtree.

  # Question Two: Path to Leaves

  Given the `root` of a binary tree, return all root to leaf paths in any order.

  ## Examples

  ### Example 1

  ![](./images/q1_ex1.png)

  Input: `root = [1, 2, 2, 3, 5, 6, 7]` *What traversal method is this?*

  Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

  ### Example 2

  ![](./images/q1_ex3.png)

  Input: `root = [10, 9, 7, 8]`

  Output: [[10, 7], [10, 9, 8]]

</details>

#### Starter Code for Question 2

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right
def bt_path(root: TreeNode):
    if not root:
        return []
    
    result = []
    
    def dfs(node, current_path):
        # Add current node to path
        current_path.append(node.val)
        
        # If it's a leaf node, add the path to result
        if not node.left and not node.right:
            result.append(current_path[:])  # Make a copy of the path
        else:
            # Recursively explore left and right subtrees
            if node.left:
                dfs(node.left, current_path)
            if node.right:
                dfs(node.right, current_path)
        
        # Backtrack: remove current node from path
        current_path.pop()
    
    dfs(root, [])
    return result

# Test cases
if __name__ == "__main__":
    # Test case 1: Simple tree
    #     1
    #    / \
    #   2   3
    #      / \
    #     4   5
    root1 = TreeNode(1)
    root1.left = TreeNode(2)
    root1.right = TreeNode(3)
    root1.right.left = TreeNode(4)
    root1.right.right = TreeNode(5)
    
    print("Test case 1:")
    print(bt_path(root1))  # Expected: [[1, 2], [1, 3, 4], [1, 3, 5]]
    
    # Test case 2: Single node
    root2 = TreeNode(1)
    print("Test case 2:")
    print(bt_path(root2))  # Expected: [[1]]
    
    # Test case 3: Linear tree
    root3 = TreeNode(1)
    root3.right = TreeNode(2)
    root3.right.right = TreeNode(3)
    
    print("Test case 3:")
    print(bt_path(root3))  # Expected: [[1, 2, 3]]
```

The recursion works with:

Base case: When we reach a leaf node (no left or right children), we add the current path to results
Recursive case: For internal nodes, we recursively explore left and right children
Backtracking: After exploring a subtree, we remove the current node from the path so it doesn't affect other paths

This finds all root-to-leaf paths in the binary tree.

  # Question Three: Missing Number in Range
 
  You are given a list containing `n` integers in the range `[0, n]`. Return a list of numbers that are missing from the range `[0, n]` of the array. If there is no missing number, return -1. Note, all the integers in the list may not be unique.
  
  ## Examples

  ### Example 1

  Input: `lst = [0, 2]`

  Output: [1]

  ### Example 2

  Input: `lst = [5, 0, 1]`

  Output: [2, 3, 4]

  ### Example 3

  Input: `lst = [6, 8, 2, 3, 5, 7, 0, 1, 10]`

  Output: [4, 9]

</details>

#### Starter Code for Question 3


```python
from typing import List

def missing_num(nums: List) -> int:
    n = len(nums)
    # Expected sum of numbers from 0 to n
    expected_sum = n * (n + 1) // 2
    # Actual sum of numbers in the array
    actual_sum = sum(nums)
    # The difference is the missing number
    return expected_sum - actual_sum

# Alternative implementation using XOR (bit manipulation)
def missing_num_xor(nums: List) -> int:
    n = len(nums)
    result = n  # Start with n since we have numbers 0 to n
    
    for i in range(n):
        result ^= i ^ nums[i]  # XOR with index and array element
    
    return result

# Alternative implementation using set
def missing_num_set(nums: List) -> int:
    n = len(nums)
    num_set = set(nums)
    
    for i in range(n + 1):
        if i not in num_set:
            return i

# Test cases
if __name__ == "__main__":
    # Test case 1: Missing number in middle
    nums1 = [3, 0, 1]
    print(f"Test case 1: {nums1}")
    print(f"Missing number: {missing_num(nums1)}")  # Expected: 2
    print(f"XOR method: {missing_num_xor(nums1)}")
    print(f"Set method: {missing_num_set(nums1)}")
    print()
    
    # Test case 2: Missing number at end
    nums2 = [0, 1]
    print(f"Test case 2: {nums2}")
    print(f"Missing number: {missing_num(nums2)}")  # Expected: 2
    print()
    
    # Test case 3: Missing number at beginning
    nums3 = [1, 2]
    print(f"Test case 3: {nums3}")
    print(f"Missing number: {missing_num(nums3)}")  # Expected: 0
    print()
    
    # Test case 4: Larger array
    nums4 = [9, 6, 4, 2, 3, 5, 7, 0, 1]
    print(f"Test case 4: {nums4}")
    print(f"Missing number: {missing_num(nums4)}")  # Expected: 8
    print()
    
    # Test case 5: Single element array
    nums5 = [1]
    print(f"Test case 5: {nums5}")
    print(f"Missing number: {missing_num(nums5)}")  # Expected: 0
```

How it works:

If we have an array of length n, it should contain numbers from 0 to n (total n+1 numbers)
The expected sum of numbers 0 to n is: n × (n + 1) / 2
We calculate the actual sum of the given array
The difference gives us the missing number

Time complexity: O(n) - single pass through the array
Space complexity: O(1) - constant extra space
I also included two alternative approaches:

XOR method: Uses bit manipulation property that x ^ x = 0
Set method: Creates a set and checks which number is missing

The mathematical approach is typically the most efficient and elegant solution for this problem.
